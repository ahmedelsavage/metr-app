```python
import easyocr

reader = easyocr.Reader(['en', 'ar'])
```

```python
result = reader.readtext('../images/testocr.png')
```

```python
for detection in result:
    text = detection[1]
    confidence = detection[2]
    print(f"Detected text: {text}")
    print(f"Confidence: {confidence}")
```

```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(use_textline_orientation=True, lang="ar")
```

```python
img_path = "../../images/arabic.png"
result = ocr.predict(img_path)

for res in result:
    res.print()
    res.save_to_img("../../images/testocr_output.png")
    res.save_to_json("../../images/testocr_output.json")
```
